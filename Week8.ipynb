Setup

```python
# Week 8 - Diffie-Hellman Key Exchange Simulation
# Partner's Name: SYED
print("Diffie-Hellman Key Exchange Simulation Started")
```

Select public values

```python
# Step 1: Publicly agreed values (p: prime number, g: base)
p = 23  # Public prime number
g = 5   # Public base (primitive root modulo p)

print(f"Public values:\np (prime) = {p}\ng (base) = {g}")
```

Private Key and Public Key

```python
# Step 2: Choose your private key (a) and calculate your public key (A)
a = 6  # Your private key (secret)
A = pow(g, a, p)  # A = g^a mod p

print(f"Your private key (a) = {a}")
print(f"Your public key (A = g^a mod p) = {A}")
```

Receive Partner's Public Key and Compute Shared Secret

```python
# Step 3: Receive partner's public key (B)
B = 19  # Example: SYED sends you their public key B

# Step 4: Compute shared secret (s = B^a mod p)
s = pow(B, a, p)

print(f"Received partner's public key (B) = {B}")
print(f"Computed shared secret (s = B^a mod p) = {s}")
```
